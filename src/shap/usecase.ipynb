```python
from rdkit import Chem
from rdkit.Chem import AllChem, Draw
from rdkit.Chem.Draw import IPythonConsole
import pandas as pd
import numpy as np
import sys
from pathlib import Path
ROOT_DIR = Path.cwd().parents[1]
smiles_df = pd.read_csv(ROOT_DIR / "datasets" / "smiles.csv",
                         header=None, names=["Drug", "SMILES"])
target_bits = [295, 127]
for bit in target_bits:
    print(f"\nDrugs with ECFP6 bit {bit}:")
    for _, row in smiles_df.iterrows():
        mol = Chem.MolFromSmiles(row["SMILES"])
        if mol is None:
            continue
        fp = AllChem.GetMorganFingerprintAsBitVect(mol, radius=6, nBits=1024)
        if fp[bit] == 1:
            print(f"  {row['Drug']}")
labels_df = pd.read_csv(ROOT_DIR / "datasets" / "labels.csv")
```

```python
methotrexate_pairs = labels_df[
    ((labels_df['drug_a_name'] == 'METHOTREXATE') | 
     (labels_df['drug_b_name'] == 'METHOTREXATE')) & 
    (labels_df['synergy'] >= 30)
]
print(methotrexate_pairs[['drug_a_name', 'drug_b_name', 'cell_line', 'synergy']].drop_duplicates())
```

```python
sorafenib_pairs = labels_df[
    ((labels_df['drug_a_name'] == 'SORAFENIB') | 
     (labels_df['drug_b_name'] == 'SORAFENIB')) & 
    (labels_df['synergy'] >= 30)
]
print(sorafenib_pairs[['drug_a_name', 'drug_b_name', 'cell_line', 'synergy']].drop_duplicates())
```

```python
top10_features = {
    'A': [127, 295, 267, 377, 702, 767],  
    'B': [767, 435]}
drug_bits = {}
for _, row in smiles_df.iterrows():
    mol = Chem.MolFromSmiles(row["SMILES"])
    if mol is None:
        continue
    fp = AllChem.GetMorganFingerprintAsBitVect(mol, radius=6, nBits=1024)
    drug_bits[row["Drug"]] = set(i for i in range(1024) if fp[i] == 1)
all_bits = set(top10_features['A'] + top10_features['B'])
for bit in sorted(all_bits):
    drugs = [drug for drug, bits in drug_bits.items() if bit in bits]
    print(f"Bit {bit}: {drugs}")
```
